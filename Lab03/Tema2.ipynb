```python
#EX 1

import numpy as np

n = 100_000

X = np.random.randint(1, 7, n)
Y = np.random.randint(1, 7, n)

A = (X + Y >= 10)
B = (X % 2 == 0)
C = (B | (Y % 2 == 0))
D = (X + Y == 8)
E = (X % 2 == 0) & (Y % 2 == 0)

p1 = np.mean(A[B])
p2 = np.mean(A[C])
p3 = np.mean(E[D])


print(f"1. P(A | B): {p1:.4f} = {p1*100:.2f}%")
print(f"   Valoare exactă: {2/9:.4f} = {2/9*100:.2f}%\n")

print(f"2. P(A | C): {p2:.4f} = {p2*100:.2f}%")
print(f"   Valoare exactă: {5/27:.4f} = {5/27*100:.2f}%\n")

print(f"3. P(E | D): {p3:.4f} = {p3*100:.2f}%")
print(f"   Valoare exactă: {3/5:.4f} = {3/5*100:.2f}%")

```

```python
# EX 3

import pandas as pd

date = {
    "Traseu": ["Urban", "Urban", "Off-road", "Off-road"],
    "Model": ["A", "B", "A", "B"],
    "Fara_defectiuni": [85, 19, 6, 65],
    "Testate": [90, 20, 10, 80]
}

df = pd.DataFrame(date)

df["Rata_succes"] = (
    df["Fara_defectiuni"] / df["Testate"] * 100
)

print("Ratele pe tipuri de traseu:")
print(df)


global_df = df.groupby("Model")[["Fara_defectiuni", "Testate"]].sum()

global_df["Rata_succes"] = (
    global_df["Fara_defectiuni"] / global_df["Testate"] * 100
)

print("\nRatele globale:")
print(global_df)
```
